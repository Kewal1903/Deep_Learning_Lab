## Load data



```python
from sklearn.datasets import load_iris

iris = load_iris()
X = iris.data
y = iris.target
```

## Preprocess data



```python
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.utils import to_categorical

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

y_categorical = to_categorical(y)
```

## Split data



```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y_categorical, test_size=0.2, shuffle=True, random_state=42)

print(f"Shape of X_train: {X_train.shape}")
print(f"Shape of X_test: {X_test.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of y_test: {y_test.shape}")
```

## Build and train model 1



```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model1 = Sequential([
    Dense(6, activation='relu', input_shape=(X_train.shape[1],)),
    Dense(y_train.shape[1], activation='sigmoid')
])

model1.compile(optimizer='adam',
               loss='categorical_crossentropy',
               metrics=['accuracy'])

model1.fit(X_train, y_train, epochs=50, batch_size=10)
```

## Evaluate model 1



```python
model1.summary()
```

```python
loss1, accuracy1 = model1.evaluate(X_test, y_test, verbose=0)
print(f"Model 1 Test Loss: {loss1:.4f}")
print(f"Model 1 Test Accuracy: {accuracy1:.4f}")
```

## Build and train model 2



```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model2 = Sequential([
    Dense(10, activation='relu', input_shape=(X_train.shape[1],)),
    Dense(6, activation='relu'),
    Dense(y_train.shape[1], activation='sigmoid')
])

model2.compile(optimizer='adam',
               loss='categorical_crossentropy',
               metrics=['accuracy'])

model2.fit(X_train, y_train, epochs=50, batch_size=10)
```

## Evaluate model 2


```python
model2.summary()
```

```python
loss2, accuracy2 = model2.evaluate(X_test, y_test, verbose=0)
print(f"Model 2 Test Loss: {loss2:.4f}")
print(f"Model 2 Test Accuracy: {accuracy2:.4f}")
```

## Model Comparison and Insights

```python
print("\n--- Model Comparison ---")
print(f"Model 1 Test Accuracy: {accuracy1:.4f}")
print(f"Model 2 Test Accuracy: {accuracy2:.4f}")
print(f"Model 1 Test Loss: {loss1:.4f}")
print(f"Model 2 Test Loss: {loss2:.4f}")
```

## Summary:

###Key Findings


*   Model 1, with one hidden layer (6 neurons), achieved a test accuracy of approximately 0.8667 and a test loss of approximately 0.4458.
*   Model 2, with two hidden layers (10 and 6 neurons), achieved a test accuracy of approximately 0.9667 and a test loss of approximately 0.1432.
*   Model 2 performed better than Model 1 on the test set, showing higher accuracy and lower loss.

### Insights

*   The increased complexity of Model 2's architecture, with an additional hidden layer and more neurons, likely allowed it to capture more complex patterns in the data, leading to improved performance on the test set.
*   Further experimentation with different numbers of layers, neurons per layer, activation functions, optimizers, and epochs could potentially further optimize the model's performance.
